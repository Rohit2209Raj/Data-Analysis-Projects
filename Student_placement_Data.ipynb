# Feature Engineering


```python
df_cleaned=df.copy()
```

```python
df_cleaned.head()
```

```python
df_cleaned.drop_duplicates(inplace=True)
```

```python
df_cleaned.columns
```

```python
df['Gender'].value_counts()
```

```python
df_cleaned['Gender']=df_cleaned['Gender'].map({'Male':0,'Female':1})
```

```python
df_cleaned
```

```python
df_cleaned['Placement_Offer'].value_counts()
```

```python
df_cleaned['Placement_Offer']=df_cleaned['Placement_Offer'].map({'Yes':1,'No':0})
```
