```python
# Now we will check for abnormalities using graphs
```

```python
numerical_columns=['Age', 'CGPA', 'Internships_Count',
       'Projects_Count', 'Certifications_Count', 'Technical_Skills_Score_100',
       'Communication_Skills_Score_100', 'Aptitude_Test_Score_100',
       'Salary_Offered_USD']
```

```python
for col in numerical_columns:
  plt.figure(figsize=(8,6))
  sns.histplot(df[col],kde=True,bins=20)
```

```python
# The histplot does not incdicate any abnormailites for lets also check it using boxplot
```

```python
for col in numerical_columns:
  plt.figure(figsize=(6,4))
  sns.boxplot(df[col])
```

```python
# we don't notice any abnormailites
```
